```python
import pandas as pd
from sqlalchemy import create_engine, Table, Column, String, MetaData, Date, BigInteger

url_symbols = "https://raw.githubusercontent.com/datasets/s-and-p-500-companies/master/data/constituents.csv"
df_symbols = pd.read_csv(url_symbols)  #csv->df
# print(type(df_symbols["Symbol"]))

df_symbols = df_symbols.rename(columns={
    "Symbol": "symbol",
    "Security": "company",
    "GICS Sector": "sector",
    "GICS Sub-Industry": "industry",
    "Headquarters Location":"headquarters_location",
    "Date added":"date_added",
    "CIK":"cik",
    "Founded":"founded"
})
df_symbols["symbol"] = df_symbols["symbol"].str.replace(".", "-", regex=False)  #some symbol (eg. BRK.B) cannot read by yahoo, change . to - can fix


db_engine = create_engine("postgresql+psycopg2://postgres:Admin1234@localhost:5432/bootcamp_2510p")

metadata = MetaData()

symbols_schema = Table(
  "symbols",
  metadata,
  Column("symbol", String(20), primary_key=True, nullable=False), #BigInteger=Long
  Column("company", String(255), nullable=False),
  Column("sector", String(100), nullable=False),
  Column("industry", String(255), nullable=False),
  Column("headquarters_location",String(255)),
  Column("date_added", Date),
  Column("cik",BigInteger),
  Column("founded",String(50))
)

metadata.create_all(db_engine)

df_symbols.to_sql("symbols", db_engine, index=False, if_exists="append") #if use if_exists"replace", pandas generate it own table, regardless to schema


```
