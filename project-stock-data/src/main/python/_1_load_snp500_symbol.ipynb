```python
import pandas as pd
from sqlalchemy import create_engine, Table, Column, MetaData, Integer, String, Date, Float, BigInteger, ForeignKey

url_symbols = "https://raw.githubusercontent.com/datasets/s-and-p-500-companies/master/data/constituents.csv"
df_symbols = pd.read_csv(url_symbols)  #csv->df
print(type(df_symbols["Symbol"]))

db_engine = create_engine("postgresql+psycopg2://postgres:Admin1234@localhost:5432/bootcamp_2510p")

metadata = MetaData()

symbol_schema = Table(
  "symbols",
  metadata,
  Column("Symbol", String(20), primary_key=True, nullable=False),
  Column("Security", String(50)),
  Column("GICS Sector", String(50)),
  Column("GICS Sub-Industry", String(50)),
  Column("Headquarters Location", String(50)),
  Column("Date added", Date),
  Column("CIK", BigInteger),
  Column("Found", BigInteger)
)

metadata.create_all(db_engine)

df_symbols.to_sql("symbols", db_engine, index=False, if_exists="replace")

```
